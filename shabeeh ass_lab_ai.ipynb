```python
students = []
for i in range(5):
    print("enter detail of student")
    name =input("enter student name:")
    roll_number = input("enter student rollnumber:")
    python_marks= int(input("enter python number:"))
    ai_marks = int(input("enter ai number:"))
    math_marks = int(input("enter the math number:"))
    student = {
        "name": name,
        "roll_no": roll_number,
        "Python": python_marks,
        "Artificial Intelligence": ai_marks,
        "Mathematics": math_marks
    }
    students.append(student)

for student in students:
    print(student)

def cal_result(student):
    total = student["Python"]+student["Artificial Intelligence"]+student["Mathematics"]
    percentage = (total/300)*100
    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"
    return total, percentage, grade

for student in students:
    total, percentage, grade = cal_result(student)
    student["Total"] = total
    student["Percentage"] = percentage
    student["Grade"] = grade
    print("\nStudent Name:", student["name"])
    print("Roll Number:", student["roll_no"])
    print("Total Marks:", total, "/ 300")
    print("Percentage:", round(percentage, 2), "%")
    print("Grade:", grade)

top_student = max(students, key=lambda student: student["Percentage"])

print("\nStudent with Highest Percentage")
print("Name:", top_student["name"])
print("Roll Number:", top_student["roll_no"])
print("Percentage:", round(top_student["Percentage"], 2), "%")
print("Grade:", top_student["Grade"])

with open("results.txt", "w") as file:
    file.write("STUDENT RESULT MANAGEMENT SYSTEM\n")

    for student in students:
        file.write("\nStudent Name: " + student["name"] + "\n")
        file.write("Roll Number: " + student["roll_no"] + "\n")
        file.write("Python Marks: " + str(student["Python"]) + "\n")
        file.write("AI Marks: " + str(student["Artificial Intelligence"]) + "\n")
        file.write("Mathematics Marks: " + str(student["Mathematics"]) + "\n")
        file.write("Total Marks: " + str(student["Total"]) + "/300\n")
        file.write("Percentage: " + str(round(student["Percentage"], 2)) + "%\n")
        file.write("Grade: " + student["Grade"] + "\n")
        file.write("-" * 30 + "\n")

    file.write("\nStudent with Highest Percentage\n")
    file.write("Name: " + top_student["name"] + "\n")
    file.write("Roll Number: " + top_student["roll_no"] + "\n")
    file.write("Percentage: " + str(round(top_student["Percentage"], 2)) + "%\n")
    file.write("Grade: " + top_student["Grade"] + "\n")

print("All results saved successfully in results.txt")

```
